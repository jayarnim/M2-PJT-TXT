# packages

```python
from pathlib import Path
from IPython.display import clear_output
import pickle
import numpy as np
import pandas as pd
from pymyproject.bcpd import bcpd
import arviz as az
```

# data

```python
PATH = "./result/factor/scores.pkl"
with open(file=PATH, mode="rb") as f:
    scores = pickle.load(f)
```

# constants

```python
N_CPS = range(1,6)
DRAWS = 10000
TUNE = 5000
TARGET_ACCEPT = 0.90
CHAINS = 4
SEED = 42
```

# factor 1

```python
FACTOR = 1
DATA = scores[f"f{FACTOR}"].values
SAVE_DIR = f"./result/bcpd/f{FACTOR}/"

for k in N_CPS:
    print(f"CPS {k} START ==========")

    # INIT. MODEL ==========
    kwargs = dict(
        data=DATA, 
        n_cps=k, 
        seed=SEED,
    )
    model = bcpd.BayesianChangePointDetection(**kwargs)

    # TRAIN ==========
    kwargs = dict(
        draws=DRAWS, 
        tune=TUNE,
        target_accept=TARGET_ACCEPT,
        chains=CHAINS,
    )
    trace = model.fit(**kwargs)

    # PREDICT ==========
    ppc = model.predict(trace)
    trace.extend(ppc)

    # SAVE ==========
    PATH = SAVE_DIR + f"cps{k}.nc"
    trace.to_netcdf(PATH)

    clear_output(wait=False)

print(f"FACTOR {FACTOR} FINISHED")
```
