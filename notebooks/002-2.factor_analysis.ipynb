# packages

```python
from pathlib import Path
import pickle
import pandas as pd
import matplotlib.pyplot as plt
from pymyproject.factor import analysis
```

# data

```python
PATH = "./result/factor/lls.pkl"
with open(file=PATH, mode="rb") as f:
    lls = pickle.load(f)
```

```python
PATH = "./result/factor/factor/"
folder = Path(PATH)

result = dict()

for i, PATH in enumerate(folder.glob("*.pkl")):
    with open(file=PATH, mode="rb") as f:
        result[i+1] = pickle.load(f)
```

# log likelihood

```python
FIGSIZE = (8,5)

kwargs = dict(
    lls=lls,
    figsize=FIGSIZE,
)

analysis.lls.main(**kwargs)
```

# k=1

```python
NUM_FACTORS = 1
data  = result[NUM_FACTORS]
```

```python
LOADINGS = data["loadings"]
FIGSIZE = (8,10)

kwargs = dict(
    loadings=LOADINGS,
    figsize=FIGSIZE,
)

analysis.loadings.main(**kwargs)
```

```python
SCORES = data["scores"]

kwargs = dict(
    scores=SCORES,
)

analysis.scores.main(**kwargs)
```
